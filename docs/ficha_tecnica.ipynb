# Ficha Técnica do Dataset

## Gait in Aging and Disease Database

| Item | Descrição |
|---|---|
| **Nome** | Gait in Aging and Disease Database |
| **Fonte** | PhysioNet |
| **Versão** | 1.0.0 |
| **DOI** | 10.13026/C2C889 |
| **Acesso** | https://physionet.org/content/gaitdb/1.0.0/ |
| **Licença** | Open Data Commons Attribution License v1.0 (ODC-By 1.0) |
| **Participantes** | 15 participantes: 5 jovens saudáveis, 5 idosos saudáveis e 5 indivíduos com Parkinson |
| **Registros processados** | 9.144 intervalos de passada |
| **Formato original** | Arquivos `.txt` com duas colunas numéricas |
| **Variáveis principais** | Tempo da medição e intervalo de passada, ambos em segundos |

---

## Variáveis utilizadas

| Variável | Descrição |
|---|---|
| `participante_id` | Identificador do participante |
| `grupo` | Jovem, idoso ou Parkinson |
| `idade` | Idade do participante quando disponível |
| `tempo_(s)` | Tempo da medição em segundos |
| `intervalo_passada_(s)` | Intervalo entre passadas em segundos |

Para os participantes com Parkinson, a idade individual não é informada nos arquivos utilizados. Por isso, esses valores foram mantidos como **nulos**.

---

## Uso clínico e Machine Learning

**Aplicação potencial:** análise quantitativa de alterações da marcha associadas ao envelhecimento e à doença de Parkinson.

**Pergunta de pesquisa:** 

> Existem diferenças na duração e na variabilidade dos intervalos de passada entre jovens saudáveis, idosos saudáveis e indivíduos com doença de Parkinson?

**Entrada:** características extraídas dos intervalos de passada.

**Alvo potencial:** grupo do participante (`jovem`, `idoso` ou `parkinson`).

---

## Riscos de privacidade

O dataset não apresenta identificadores pessoais diretos, como nome ou endereço. Entretanto, contém algumas informações demográficas e clínicas, como idade e grupo de saúde.

O risco de reidentificação é considerado limitado, mas não inexistente.

## Limitações

- Apenas **15 participantes**, sendo 5 por grupo.
- O número de passadas é diferente entre os grupos.
- O protocolo de coleta dos participantes com Parkinson foi diferente do utilizado nos participantes saudáveis.
- As passadas de um mesmo participante são medidas repetidas e **não devem ser consideradas observações independentes**.
- As idades individuais dos participantes com Parkinson não estão disponíveis nos arquivos utilizados.
